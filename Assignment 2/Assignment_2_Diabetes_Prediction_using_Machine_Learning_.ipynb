**3.Separate the dataset into:**

*a.   Features (X)*

*b.  Target Variable (y)*

```
X = data.drop(['Outcome'], axis =1) #input columns excluding outcome
y = data['Outcome']
```

```
X.head()
```

**4. Perform a Train-Test Split**

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15)
```

```
X_train.shape
```

```
X_test.shape
```

**Part 2 — Machine Learning Model**

**Logistic Regeression**

```
# Import libraries
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```
logistic_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
                                max_iter = 1000,
                                random_state=42
                            ))
                            ])
```

**Decision Tree Classifier**

```
from sklearn.tree import DecisionTreeClassifier

decison_tree_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", DecisionTreeClassifier(
        max_depth = 5,
        random_state=42
    ))
])
```

**Random Forest Classifier**

```
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", RandomForestClassifier(
        n_estimators = 200,
        max_depth=8,
        random_state=42
    ))
])
```

**Train all the models**

```
models = {
    "Logestic Regression" : logistic_model,
    "Decision Tree": decison_tree_model,
    "Random Forest": random_forest_model
}

for name, model in models.items():
  model.fit(X_train, y_train)
  print(f"{name} trained successfully.")
```

**Generate Predictions**

```
predictions = {}
for name, model in models.items():
  predictions[name] = model.predict(X_test)
```

**Evaluvate Different Models**

```
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score )


results =[]
for name, model in models.items():
  y_pred = model.predict(X_test)

  accuracy = accuracy_score(y_test,y_pred)
  precision = precision_score(y_test, y_pred)
  recall = recall_score(y_test, y_pred)
  f1 = f1_score(y_test,y_pred)

  results.append({
      "Model":name,
      "Accuracy":accuracy,
      "Precision":precision,
      "Recall":recall,
      "F1-Score":f1
  })


results_df = pd.DataFrame(results)
print(results_df)
```

# **From this Random Forest is the Best model**

**Part 3 — Model Training**

 ***Fit the selected model using the training dataset.***

```
random_forest_model.fit(X_train, y_train)
print(f"Random forest model trained successfully.")
```

***Generate predictions on the training dataset***

```
y_train_pred = random_forest_model.predict(X_train)
print("First 10 Training Predictions")
print(y_train[:10])
```

***Evaluate the model's training performance.***

```
accuracy = accuracy_score(y_train,y_train_pred)
precision = precision_score(y_train, y_train_pred)
recall = recall_score(y_train, y_train_pred)
f1 = f1_score(y_train,y_train_pred)

print("Training data Performance")
print(f"Accuracy : {accuracy}")
print(f"Precision : {precision}")
print(f"Recall : {recall}")
print(f"F1-Score : {f1}")
```

***Generate predictions on the Test dataset and Evaluating the Performance***

```
y_test_pred = random_forest_model.predict(X_test)
accuracy_test = accuracy_score(y_test,y_test_pred)
precision_test = precision_score(y_test, y_test_pred)
recall_test = recall_score(y_test, y_test_pred)
f1_test = f1_score(y_test,y_test_pred)

print("Test data Performance")
print(f"Accuracy : {accuracy_test}")
print(f"Precision : {precision_test}")
print(f"Recall : {recall_test}")
print(f"F1-Score : {f1_test}")
```

***Check for potential overfitting or underfitting by comparing training and testing performance.***

**Comparison Table**

```
performance_comparsion = pd.DataFrame({
    "Metric" : [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
    ],
    "Training":[
        accuracy,
        precision,
        recall,
        f1
    ],
    "Testing":[
      accuracy_test,
      precision_test,
      recall_test,
      f1_test
    ]

})

performance_comparsion
```

```
accuracy_difference = accuracy - accuracy_test
f1_difference = f1-f1_test
if accuracy > 0.90 and accuracy_difference > 0.10 :
  print("Potential overfitting")
elif accuracy<0.70 and accuracy_test <0.70:
  print("Potential underfitting")
else:
  Print("No potential Overfitting or Underfitting")
```
