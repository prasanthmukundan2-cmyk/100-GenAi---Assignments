# **Part 1 - Data Understanding and Exploration**

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**1. Load Dataset**

```python
data = pd.read_csv('/content/test_zomato.csv')
data.head()
```

**2. Explore the Dataset**


1.   Size of the dataset
2.   Columns in the dataset

1.   Datatypes of each columns
2.   Information about the dataset


1.   Missing values








```python
# Size/Dimensions of the dataset
data.shape

# Column names in the dataset
data.columns

# Datatypes of each columns
data.dtypes

# Information abount the dataset
data.info()

# Missing values
data.isnull().sum()
```

**`3. Duplicate Records`**

```python
# Checks whether duplicate exist or not
data.duplicated().sum()

#Drop duplicates
data.drop_duplicates(inplace= True)
```

**4. Convert the "Average Cost for two" column into an appropriate numeric data type**

```python
# Check the datatype of the column
data["Average Cost for two"].dtype

# Convert it into numeric datatype
data["Average Cost for two"] = pd.to_numeric(data["Average Cost for two"], errors="coerce")

data["Average Cost for two"].dtype
```

**5. Statistical Summary**

```python
# summarizes numeric columns
data.describe()

# summarizes all columns
data.describe(include="all")
```

**6. Answer the Questions**

```python
# Unique Cities
data["City"].unique()

# City with highest number of Restaurants
data["City"].value_counts().head(1)

# Most common Cuisine
data["Cuisines"].value_counts().head(1)
```

# **Part 2 - Data Analysis**


```python
# 1.Find the average restaurant rating.
data["Aggregate rating"].mean()

# 2. Identify the restaurant with the highest number of votes.
data.loc[data["Votes"].idxmax()]

# 3. Find the average cost for two across all restaurants.
data["Average Cost for two"].mean()

# 4. Compare the average ratings of restaurants that:
  # a. Have Online Delivery
  # b. Do Not Have Online Delivery
data.groupby("Has Online delivery")["Aggregate rating"].mean()

# 5. Compare the average ratings of restaurants that:
  # a. Have Table Booking
  # b. Do Not Have Table Booking
data.groupby("Has Table booking")["Aggregate rating"].mean()

# 6. Identify the top 10 cities with the highest number of restaurants.
data["City"].value_counts().head(10)
```

# **Part 2 - Data Visualization**


**1. Bar Chart of Top 10 cities with the highest number of restaurants** **bold text**

```python
plt.figure(figsize=(20,5))
top10 = data['City'].value_counts().head(10)
plt.bar(top10.index, top10.values)
plt.xlabel('City')
plt.ylabel('Count')
plt.title('Top 10 cities with the highest number of restaurants')
plt.show()
```

**2. Histogram for Distribution of Aggregate Ratings.**

```python
plt.figure(figsize=(8,5))
sns.histplot(data["Aggregate rating"], bins=20)
plt.title('Distribution of Aggregate Ratings')
plt.show()
```

**3. Bar Chart for Number of restaurants offering Online Delivery (Yes/No).**

```python
plt.figure(figsize=(10,5))
OnlineDelivery = data["Has Online delivery"].value_counts()
plt.bar(OnlineDelivery.index, OnlineDelivery.values)
plt.xlabel('Has online Delivery')
plt.ylabel('No. of Cities')
plt.title('Restaurants offering Online Delivery')
plt.show()
```

**4. Bar Chart for Top 10 most common cuisines.**

```python
plt.figure(figsize=(20,5))
top_Cuisine = data["Cuisines"].value_counts().head(10)
plt.bar(top_Cuisine.index, top_Cuisine.values)
plt.xlabel('Cuisines')
plt.ylabel('Count')
plt.title('Top 10 most common cuisines')
plt.show()
```

**5. Line Chart for Average restaurant rating by Price Range.**

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(20,5))
rating_price = data.groupby("Price range")["Aggregate rating"].mean()
plt.plot(rating_price.index, rating_price.values)
plt.xlabel('Price Range')
plt.ylabel('Average Rating')
plt.title('Avearge rating by price range')
plt.xticks(rating_price.index)
plt.grid(True)
plt.show()
```
